```python
'''Structure & Dimension Check'''
print("Shape data:", df.shape)
df.info

display(df.head())
display(df.tail())
display(df.sample(5))
```

```python
# Check missing value explicit
missing_data = pd.DataFrame({
    'total_missing' : df.isnull().sum(),
    'percentage_of_missing' : (df.isnull().sum() / len(df)) * 100
})
missing_data[missing_data['total_missing'] > 0 ]
```

```python
# Check missing value implicit
implicit_missing = ['?', '-', 'N/A', 'null', 'none', ' ']
for col in df.select_dtypes(include=['object', 'string']).columns:
    found = df[col].astype(str).str.strip().isin(implicit_missing).sum()
    if found > 0:
        print(f"column'{col}' has {found} missing value")
print("done")
```

```python
#Duplicated row
print("Total duplicated rows :", df.duplicated().sum())

#unique id
print("total unique order id :", df['Order Id'].nunique())
print("total unique order item id :", df['Order Item Id'].nunique())

```

```python
#Categorical data check 
for col in df.select_dtypes(include=['object', 'category', 'string']).columns:
    print(f"colums: '{col}' (unique: '{df[col].nunique()}')")
    print(df[col].value_counts(dropna=False).head(10))
    print("-" * 40)
```

```python
# Descriptive statistics for numerical columns
df.describe().T

# Detection of illogical values

num_cols = df.select_dtypes(include='number').columns
for col in num_cols:
    negatives = (df[col] < 0).sum()
    if negatives > 0:
        print(f"Column'{col}' has {negatives} negative value")

```
